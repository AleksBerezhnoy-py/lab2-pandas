```python
import pandas as pd
import numpy as np
```

```python
# Сотрудники в it
data = {
    'name': ['Иван', 'Мария', 'Александр', 'Анна', 'Олег', 'Елена', 'Дмитрий', 'Ольга'],
    'department': ['IT', 'IT', 'HR', 'IT', 'HR', 'Finance', 'Finance', 'IT'],
    'salary': [80000, 95000, 60000, 75000, 65000, 90000, np.nan, 72000],
    'bonus': [10000, np.nan, 5000, 8000, np.nan, 12000, 7000, np.nan],
    'hire_date': pd.to_datetime(['2020-01-15','2019-03-10','2021-07-01','2018-11-20',
                                 '2022-02-01','2017-05-15','2020-09-30','2021-12-01'])
}
df = pd.DataFrame(data)
df
```

#### Задача 2
#### Освоить методы доступа к данным и создания новых столбцов

```python
# Столбец name и salary для сотрудников из IT
df.loc[df['department'] == 'IT', ['name', 'salary']]
```

```python
# первые 3 строки и первые 2 столбца
df.iloc[:3, :2]
```

```python
# зарплата выше 70000
df[df['salary'] > 70000]
```

```python
# новый столбец total_comp
df['total_comp'] = df['salary'] + df['bonus'].fillna(0)
```

```python
# удалил столбец hire_date
df = df.drop(columns=['hire_date'])
```

#### Задача 3
#### Научиться обнаруживать и заполнять/удалять пропуски

```python
# 3.1 Количество пропусков в каждом столбце
df.isna().sum()
```

```python
# 3.2 Заполнить пропуски в salary средним значением по отделу
df['salary'] = df['salary'].fillna(df.groupby('department')['salary'].transform('mean'))
```

```python
# 3.3 Заполнить пропуски в bonus нулём
df['bonus'] = df['bonus'].fillna(0)
```

```python
# 3.4 Удалить строки, где salary всё ещё пропущен
df = df.dropna(subset=['salary'])
```

```python
df['total_comp'] = df['salary'] + df['bonus']
```

```python
df
```

#### Задача 4
#### Освоить groupby, агрегатные функции и сводные таблицы

```python
# 4.1 Средняя зарплата по каждому отделу
df.groupby('department')['salary'].mean()
```

```python
# 4.2 Для каждого отдела средняя зарплата, максимальный бонус, количество сотрудников
df.groupby('department').agg(
    avg_salary=('salary', 'mean'),
    max_bonus=('bonus', 'max'),
    employees=('name', 'count')
)
```

```python
# 4.3 Сводная таблица
df.pivot_table(index='department', values='salary', aggfunc='mean') 
```

```python
# 4.4 Отсортировать отделы по средней зарплате по убыванию
df.groupby('department')['salary'].mean().sort_values(ascending=False)
```

#### Задача 5
#### Научиться объединять DataFrame'ы с помощью merge и concat

```python
# 5.1 DataFrame managers с отделами и руководителями
managers = pd.DataFrame({
    'department': ['IT', 'HR', 'Finance'],
    'manager': ['Сергей', 'Ирина', 'Алексей']
})
managers
```

```python
# 5.2 Left join df с managers по department
df_merged = df.merge(managers, on='department', how='left')
df_merged.head(2)
```

```python
# 5.3 Добавить новую строку через concat
new_row = pd.DataFrame([{
    'name': 'Новый', 'department': 'IT', 'salary': 85000,
    'bonus': 9000, 'total_comp': 94000
}])
df_merged = pd.concat([df_merged, new_row], ignore_index=True)
```

```python
# 5.4 Сбросить индекс
df_merged = df_merged.reset_index(drop=True)
```
